```python
import pandas as pd
import numpy as np

np.random.seed(42)
records = 50

data = {
    "Order_ID": [f"ORD{1001+i}" for i in range(records)],
    "Order_Date": pd.date_range(start="2024-01-01", periods=records, freq="5D"),
    "Product": np.random.choice(
        ["Laptop", "Mouse", "Keyboard", "Monitor", "Printer"], records
    ),
    "Category": np.random.choice(
        ["Electronics", "Accessories"], records
    ),
    "Quantity": np.random.randint(1, 6, records),
    "Price": np.random.randint(500, 50000, records)
}

df = pd.DataFrame(data)

df["Total_Sales"] = df["Quantity"] * df["Price"]

total_revenue = df["Total_Sales"].sum()
total_orders = df["Order_ID"].nunique()

sales_summary = {
    "Total Revenue": total_revenue,
    "Total Orders": total_orders
}

print("Overall Sales Summary:")
print(sales_summary)

product_wise_sales = df.groupby("Product")["Total_Sales"].sum().reset_index()
print("\nProduct-wise Sales:")
print(product_wise_sales)

category_wise_sales = df.groupby("Category")["Total_Sales"].sum().reset_index()
print("\nCategory-wise Sales:")
print(category_wise_sales)

df["Month"] = df["Order_Date"].dt.to_period("M")
monthly_sales = df.groupby("Month")["Total_Sales"].sum().reset_index()

print("\nMonthly Sales:")
print(monthly_sales)

df.to_csv("week2_sales_analysis_output.csv", index=False)

print("\nFile exported as week2_sales_analysis_output.csv")
```
